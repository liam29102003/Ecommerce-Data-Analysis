```python
%sql
-- Null counts per column in customers
SELECT
  COUNT(*) AS total_rows,
  COUNT(*) - COUNT(city) AS null_city,
  COUNT(*) - COUNT(region_id) AS null_region,
  COUNT(*) - COUNT(birth_date) AS null_birth_date,
  COUNT(*) - COUNT(customer_segment) AS null_segment,
  COUNT(*) - COUNT(acquisition_channel) AS null_channel
FROM ecommerce.base.customers;
```

```python
%sql
-- Null counts in products (brand is documented as partially null)
SELECT
  COUNT(*) AS total_rows,
  COUNT(*) - COUNT(brand) AS null_brand,
  COUNT(*) - COUNT(unit_cost) AS null_cost,
  COUNT(*) - COUNT(list_price) AS null_price
FROM ecommerce.base.products;
```

```python
%sql
-- Null counts in orders (shipping_method documented as sometimes null)
SELECT
  COUNT(*) AS total_rows,
  COUNT(*) - COUNT(shipping_method) AS null_shipping_method,
  COUNT(*) - COUNT(store_id) AS null_store,
  COUNT(*) - COUNT(discount_amount) AS null_discount
FROM ecommerce.base.orders;
```
