```python
%sql
-- Monthly revenue trend
SELECT
  DATE_TRUNC('month', order_date) AS order_month,
  SUM(order_total) AS revenue,
  COUNT(DISTINCT order_id) AS orders_count
FROM ecommerce.base.orders
WHERE order_status = 'Completed'
GROUP BY 1
ORDER BY 1;
```

```python
%sql
-- Revenue by category (top-line product performance)
SELECT
  c.category_name,
  SUM(oi.item_revenue) AS revenue,
  SUM(oi.quantity) AS units_sold
FROM ecommerce.base.order_items oi
JOIN ecommerce.base.products p ON oi.product_id = p.product_id
JOIN ecommerce.base.categories c ON p.category_id = c.category_id
GROUP BY c.category_name
ORDER BY revenue DESC;
```

```python
%sql
-- Order status mix (completed vs cancelled vs returned vs pending)
SELECT order_status, COUNT(*) AS orders, 
       ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 2) AS pct_of_total
FROM ecommerce.base.orders
GROUP BY order_status;
```
