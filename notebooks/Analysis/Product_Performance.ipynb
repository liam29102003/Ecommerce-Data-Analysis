### Q5 — Which products have the highest sales volume?

```python
%sql
SELECT
  product_id, product_name, brand, category_name,
  SUM(quantity)                                   AS units_sold,
  ROUND(SUM(item_revenue), 2)                     AS revenue,
  RANK() OVER (ORDER BY SUM(quantity) DESC)       AS volume_rank
FROM ecommerce.mart.fct_sales_flat
WHERE is_completed = TRUE
GROUP BY product_id, product_name, brand, category_name
ORDER BY units_sold DESC
LIMIT 20;
```

### Q6 — Which brands generate the most revenue?

```python
%sql
SELECT
  brand,
  COUNT(DISTINCT product_id)                       AS product_count,
  SUM(quantity)                                    AS units_sold,
  ROUND(SUM(item_revenue), 2)                      AS revenue,
  ROUND(SUM(gross_profit), 2)                      AS gross_profit,
  RANK() OVER (ORDER BY SUM(item_revenue) DESC)    AS revenue_rank
FROM ecommerce.mart.fct_sales_flat
WHERE is_completed = TRUE
GROUP BY brand
ORDER BY revenue DESC;
```
