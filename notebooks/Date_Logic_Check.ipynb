```python
%sql
-- Payments recorded before their order date
SELECT p.payment_id, p.order_id, o.order_date, p.payment_date
FROM ecommerce.base.payments p
JOIN ecommerce.base.orders o ON p.order_id = o.order_id
WHERE p.payment_date < o.order_date;

-- SELECT count(*)
-- FROM ecommerce.base.payments p
-- JOIN ecommerce.base.orders o ON p.order_id = o.order_id;
```

```python
%sql
-- Returns recorded before their order date
SELECT r.return_id, r.order_id, o.order_date, r.return_date
FROM ecommerce.base.returns r
JOIN ecommerce.base.orders o ON r.order_id = o.order_id
WHERE r.return_date < o.order_date;
```

```python
%sql
-- Date range covered by orders (confirm it matches the date dimension's 2023-2025)
SELECT MIN(order_date) AS earliest_order, MAX(order_date) AS latest_order
FROM ecommerce.base.orders;
```
